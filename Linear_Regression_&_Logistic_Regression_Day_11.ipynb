Today's Objective

Understand two important ML algorithms:

Linear Regression → numerical prediction
Logistic Regression → classification
**1. Linear Regression**

Definition:
Linear Regression is a supervised ML algorithm used to predict a continuous numerical value.

Purpose:
To understand and predict how an output changes based on input features.

Important Points:

Used for regression problems.
Predicts numerical values.
Finds a relationship between input and output.
The basic idea is to fit a straight line through the data.

⭐ Remember:
Linear Regression → Predict a numerical value.

**2. Logistic Regression**

Definition:
Logistic Regression is a supervised ML algorithm mainly used for classification.

Purpose:
To predict which category/class an observation belongs to.

Important Points:

Used for classification problems.
Produces a probability between 0 and 1.
The probability is used to determine the predicted class.
Despite its name, it is mainly a classification algorithm.

⭐ Remember:
Logistic Regression → Predict a class/category.

3. Linear vs Logistic Regression
Linear Regression	Logistic Regression
Regression	Classification
Predicts numerical value	Predicts class/category
Output is continuous	Output is probability/class
⭐ Final Revision

Linear → Number
Logistic → Category
