Day 14 — Naive Bayes & K-Means Clustering



**1. Naive Bayes**

Definition:
Naive Bayes is a supervised ML algorithm based on probability and Bayes' theorem.

Purpose:
Mainly used for classification problems.

Important Points:

Uses probability to predict a class.
Assumes that features are independent of each other.
Simple and fast to train.
Commonly useful when working with classification data.

 Remember:
Naive Bayes → Probability-based classification

**2. K-Means Clustering**

Definition:
K-Means is an unsupervised ML algorithm that divides data into K groups (clusters) based on similarity.

Purpose:
Used when the data does not have predefined labels.

Important Points:

K represents the number of clusters.
Similar data points are placed in the same cluster.
Each cluster has a central point called a centroid.
Useful for discovering natural groups in data.

 Remember:
K-Means → Unsupervised → Groups similar data
