**1. Decision Tree**

Definition:
A Decision Tree is a supervised ML algorithm that makes predictions by following a series of questions or conditions.

Purpose:
Used for both classification and regression.

Important Points:

Data is divided step-by-step based on conditions.
The structure looks like a tree.
Final result is produced at the leaf node.
Easy to understand and interpret.

 Remember:
Decision Tree → Step-by-step decisions → Classification or Regression

**2. Random Forest**

Definition:
Random Forest is an ensemble ML algorithm that combines multiple Decision Trees to make a prediction.

Purpose:
To produce more reliable predictions than using a single Decision Tree.

Important Points:

Uses multiple Decision Trees.
Each tree learns from different portions/features of the data.
Their predictions are combined to produce the final result.
Can be used for classification and regression.

 Remember:
Random Forest → Multiple Decision Trees → Combined prediction

**Today's Revision**

Decision Tree → One tree
Random Forest → Many trees together
