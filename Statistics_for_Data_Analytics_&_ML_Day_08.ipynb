📊 1. Population

What is it?
The complete set of data or individuals we are interested in studying.

Remember: Population = Entire data

📋 2. Sample

What is it?
A smaller portion selected from the population.

Why use it?
Studying the entire population may be difficult or expensive, so we analyze a representative sample.

Remember: Sample = Part of population

3. Mean

What is it?
Mean is the average value of a dataset.

Role: Gives a central value representing the data.

⭐ Important for basic statistical analysis.

4. Median

What is it?
Median is the middle value when data is arranged in order.

Role: Represents the center of data and is less affected by extreme values than the mean.

⭐ Important: Very useful when data contains outliers.

5. Mode

What is it?
Mode is the value that occurs most frequently.

Role: Helps identify the most common value or category.

6. Range

What is it?
Range is the difference between the highest and lowest values.

Role: Gives a basic idea of how spread out the data is.

7. Variance ⭐⭐

What is it?
Variance measures how far data values are spread from the mean.

Role: Helps understand the variability of data.

8. Standard Deviation ⭐⭐⭐

What is it?
Standard deviation measures the typical spread of values around the mean.

Role: Helps understand whether data values are closely grouped or widely spread.

⭐ Remember: Higher standard deviation → greater spread.

📈 9. Distribution

What is it?
Distribution describes how values are spread across a dataset.

Role: Helps understand the shape and behavior of data.

🔔 10. Normal Distribution ⭐⭐⭐

What is it?
A normal distribution is a symmetric, bell-shaped distribution where values are concentrated around the center.

Role: It is an important statistical concept used in analysis and Machine Learning.

⭐ Remember: Symmetric + bell-shaped.

📉 11. Skewness ⭐⭐⭐

What is it?
Skewness describes how much a distribution is asymmetric.

Types:

Positive skew → Tail toward higher values.
Negative skew → Tail toward lower values.
Zero skew → Approximately symmetric.

⭐ Important: Skewness helps understand whether data is balanced or pulled toward one side.

12. Percentile

What is it?
A percentile indicates the position of a value relative to the rest of the dataset.

Role: Helps compare an observation with other observations.

13. Quartiles

What is it?
Quartiles divide ordered data into four sections.

Important:

Q1 → 25th percentile
Q2 → 50th percentile / Median
Q3 → 75th percentile
14. IQR ⭐⭐⭐

Full Form: Interquartile Range

What is it?
IQR represents the spread of the middle 50% of the data.

Role: It is commonly used to identify outliers.

Remember: IQR = Q3 − Q1

🚨 15. Outliers

What is it?
An outlier is an unusually high or low observation compared with the rest of the data.

Role: Outliers can represent errors, unusual events, or genuine extreme values.

⭐ Don't automatically remove outliers.

🎲 16. Probability

What is it?
Probability measures the likelihood of an event occurring.

Role: It provides the foundation for many statistical and Machine Learning concepts.

Remember: Probability ranges from 0 to 1.

🔗 17. Correlation ⭐⭐⭐

What is it?
Correlation measures the strength and direction of the relationship between two variables.

Role: Helps identify whether variables tend to move together.

Remember: Correlation ranges from -1 to +1.

18. Covariance

What is it?
Covariance indicates the direction in which two variables tend to change together.

Role: Helps understand whether variables move in the same or opposite direction.

⭐ Correlation is easier to interpret because it is standardized.

⚠️ 19. Correlation vs Causation ⭐⭐⭐

Correlation: Two variables are associated.

Causation: A change in one variable actually causes a change in another.

Important: Correlation does not automatically mean causation.
