Day 15 — Gradient Boosting & XGBoost

Today we’ll focus only on these two topics.

**1. Gradient Boosting**

Definition:
Gradient Boosting is an ensemble ML technique that builds multiple models sequentially, where each new model tries to improve the previous model's mistakes.

Purpose:
Used for classification and regression problems.

Important Points:

Models are built one after another.
Each new model focuses on previous errors.
Predictions from multiple models are combined.
Can provide strong predictive performance.

 Remember:
Gradient Boosting → Models built sequentially → Correct previous errors

**2. XGBoost**

Definition:
XGBoost (Extreme Gradient Boosting) is an optimized and powerful implementation of gradient boosting based mainly on decision trees.

Purpose:
Used for classification and regression.

Important Points:

Based on gradient boosting.
Designed for speed and efficiency.
Handles structured/tabular data very well.
Includes techniques to reduce overfitting.

 Remember:
XGBoost → Optimized Gradient Boosting → Strong performance on tabular data
