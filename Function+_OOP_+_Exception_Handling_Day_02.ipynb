```python
#1. Functions

A function is a reusable block of instructions designed to perform a specific task.
Functions help make programs organized, readable, and reusable.
Parameters are the inputs defined by a function.
Arguments are the actual values passed to the function.
The return statement sends a result back from a function.

#2. *args

*args allows a function to accept multiple positional arguments.
The received values are handled as a collection inside the function.

#3. **kwargs

**kwargs allows a function to accept multiple keyword arguments.
The received values are handled as key-value pairs.

#4. Lambda Functions

A lambda function is a small anonymous function.
It is mainly useful for short operations.
Lambda functions are commonly used with functions such as map and filter.

#5. Map

Map applies an operation to every item in a collection.
It is useful when you want to transform multiple values.

#6. Filter

Filter selects items from a collection based on a condition.
It is useful when you only want values that satisfy a particular requirement.

#7. Exception Handling

Exception handling allows a program to deal with unexpected errors.
It prevents the entire program from stopping because of a single error.
The main concepts are try, except, else, and finally.

#8. File Handling

File handling allows Python to read information from and write information to files.
It is important in Data Science because datasets are commonly stored in files.

#9. JSON

JSON is a lightweight format used to store and exchange structured data.
It is especially important when working with APIs and web applications.
JSON commonly represents information using key-value pairs and collections.

#10. OOP

Object-Oriented Programming is a programming approach based on classes and objects.
A class defines the structure and behavior of an object.
An object is an instance created from a class.
A constructor is used to initialize an object.
Attributes represent the data associated with an object.
Methods represent the actions an object can perform.

#11. Inheritance

Inheritance allows one class to acquire properties and behaviors from another class.
It helps reduce duplicate code and supports code reuse.

#12. Encapsulation

Encapsulation means keeping related data and functionality together inside an object.
It also helps control how data can be accessed or modified.
```

```python
#Functions

def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

```python
*args     → multiple positional arguments
**kwargs  → multiple keyword arguments
```

```python
#*args # single

def total(*numbers):
    return sum(numbers)

print(total(10, 20))
print(total(10, 20, 30, 40))
```

```python
#*args # multiple

def student_info(**details):
    print(details)

student_info(
    name="Aniket",
    age=22,
    role="Data Scientist"
)
```

```python
#Lambda Functions
def square(x):
    return x * x
```

```python
square = lambda x: x * x

print(square(5))
```

```python
#map()
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x**2, numbers))

print(squares)

```

```python
#filter()

numbers = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)
```

```python
#Exception Handling

try:
    x = 10 / 0

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program finished")
```
