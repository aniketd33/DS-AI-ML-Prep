```python
DAY 7 — SQL Advanced Analytics

Total Time: ~2 Hours

Today we continue SQL from yesterday. The focus is on real Data Analyst interview and job-level SQL.

🎯 Today's Agenda
Part 1 — Subqueries & CTEs | 40 min
Subquery
Nested Query
CTE
CTE vs Subquery
Part 2 — Window Functions | 50 min ⭐⭐⭐
What is a Window Function?
ROW_NUMBER
RANK
DENSE_RANK
LEAD
LAG
PARTITION BY
Part 3 — SQL Concepts | 30 min
UNION vs UNION ALL
Primary Key vs Foreign Key
Constraints
SQL NULL handling
Query execution basics

1. Subquery ⭐⭐

What is it?
A subquery is a query written inside another SQL query.

Why use it?
It allows the result of one query to be used by another query.

Where used?
Filtering, comparisons, and solving multi-step analytical problems.

Remember: Subquery = Query inside Query

2. CTE ⭐⭐⭐

Full Form: Common Table Expression

What is it?
A CTE is a temporary named result that can be used within a SQL query.

Why use it?
It makes complex queries easier to read, organize, and maintain.

Where used?
Complex data analysis and multi-step SQL transformations.

Remember: CTE = Temporary named result

3. Window Function ⭐⭐⭐

What is it?
A window function performs calculations across related rows without combining them into a single row.

Why use it?
It allows us to calculate rankings, running totals, comparisons, and previous/next values while keeping individual records.

⭐ Very important for Data Analyst interviews.

4. ROW_NUMBER

What is it?
Assigns a unique sequential number to each row.

Use: Ranking or numbering records.

Remember: ROW_NUMBER = Unique position

5. RANK

What is it?
Assigns rankings to records based on their values.

Important: Equal values receive the same rank, and the next rank can be skipped.

Remember: RANK = Ranking with gaps

6. DENSE_RANK

What is it?
Similar to RANK, but it does not leave gaps after tied values.

Remember: DENSE_RANK = Ranking without gaps

7. LEAD

What is it?
LEAD accesses a value from a following row.

Why use it?
Useful for comparing the current record with a future record.

Remember: LEAD = Look forward

8. LAG

What is it?
LAG accesses a value from a previous row.

Why use it?
Useful for comparing current data with previous data.

Remember: LAG = Look backward

9. PARTITION BY ⭐⭐⭐

What is it?
PARTITION BY divides data into separate groups for a window function.

Why use it?
It allows calculations or rankings to be performed independently within each group.

Remember: PARTITION BY = Separate into groups

10. UNION vs UNION ALL

UNION combines results and removes duplicate records.

UNION ALL combines results and keeps duplicates.

⭐ Remember:

UNION = Combine + Remove duplicates

UNION ALL = Combine + Keep duplicates

11. Constraints

What are Constraints?
Constraints are rules applied to database columns to maintain data quality and integrity.

Important constraints include:

PRIMARY KEY → Unique identification
FOREIGN KEY → Relationship between tables
NOT NULL → Value cannot be missing
UNIQUE → Values must be unique
CHECK → Value must satisfy a condition
DEFAULT → Provides a default value
```
