**Day 12 — K-Nearest Neighbors (KNN) & Support Vector Machine (SVM)**

Today we’ll focus only on these two topics.

1. K-Nearest Neighbors (KNN) **bold text**

Definition:
KNN is a supervised ML algorithm that predicts the output of a data point based on the closest data points around it.

Purpose:
Used for classification and regression.

Important Points:

Finds the K nearest data points.
For classification, nearby points help decide the class.
The value of K determines how many neighbors are considered.
Distance between data points is important.

⭐ Remember:
KNN → Look at nearby points → Make prediction

**2. Support Vector Machine (SVM)**

Definition:
SVM is a supervised ML algorithm that finds a boundary (decision boundary) to separate different classes.

Purpose:
Mainly used for classification, and can also be used for regression.

Important Points:

Separates classes using a decision boundary.
Tries to create the best separation between classes.
The closest important data points are called support vectors.
Useful for classification problems.

⭐ Remember:
SVM → Find the best boundary → Separate classes
